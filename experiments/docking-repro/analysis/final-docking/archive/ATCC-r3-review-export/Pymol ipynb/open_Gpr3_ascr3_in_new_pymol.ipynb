# Gpr3 WT / ascr3 — ATCC full-length r3 review

Run the next cell once. It starts a new local PyMOL app instance and loads the r3 full-length prepared receptor plus every independently numbered pose. Report decisions only by `ATCCFLR3-*` stable ID.


```python
from pathlib import Path
import subprocess
import time

job_dir = (Path.cwd() / "../Gpr3_ascr3").resolve()
if not job_dir.is_dir():
    raise FileNotFoundError(f"Expected job folder: {job_dir}")

pymol_executable = Path("/opt/homebrew/bin/pymol")
if not pymol_executable.exists():
    raise FileNotFoundError(f"PyMOL executable not found: {pymol_executable}")
pml_file = job_dir / "load_all_poses_pearl.pml"
process = subprocess.Popen(
    [str(pymol_executable), "-I", str(pml_file)],
    cwd=job_dir,
    start_new_session=True,
)
time.sleep(2)
if process.poll() is not None:
    raise RuntimeError(f"PyMOL exited during launch with code {process.returncode}")
print("Started a new PyMOL process")
print("PID:", process.pid)
print("Job:", "Gpr3_ascr3")
print("Expected poses:", 20)
print("Script:", pml_file)
```
